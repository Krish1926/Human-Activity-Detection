```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split

human_fall = pd.read_csv('All Users Combined.csv')
human_fall
```

```python
print(human_fall.info())
print("\n",human_fall['activity'].value_counts())
```

### Data Cleaning

```python
print("Missing Values in the Human Fall data:\n")
human_fall.isnull().sum()
```

```python
# Checking duplicates count per column
duplicates_per_column = human_fall.apply(lambda col: col.duplicated().sum())

print("Number of duplicate values in each column:")
print(duplicates_per_column)
```

```python
# Checking duplicate rows
print("Number of Duplicate Rows in the Human Fall data: ",human_fall.duplicated().sum())
print(human_fall.duplicated())
human_fall = human_fall.drop_duplicates()
print("\nNumber of Duplicates after dropping from the data:\n",human_fall.duplicated())
```

```python
# Keep 'downstaires' as it is
downstaires_df = human_fall[human_fall['activity'] == 'downstaires']

# Classes to downsample
classes_to_downsample = ['Sitting', 'Walking', 'Running', 'upstaires', 'Standing']

# Downsample with 4500
downsampled_dfs = [
    human_fall[human_fall['activity'] == cls].sample(n=4500, random_state=41)
    for cls in classes_to_downsample
]

# Combine both downsampled classes with 'downstaires'
HF_balanced = pd.concat(downsampled_dfs + [downstaires_df], ignore_index=True)
HF_balanced = HF_balanced.sample(frac=1, random_state=42).reset_index(drop=True)
print(HF_balanced['activity'].value_counts())
```

```python
from sklearn.preprocessing import LabelEncoder
import warnings
warnings.filterwarnings('ignore')

# Encode target labels
human_le = LabelEncoder()
HF_balanced['activity']= human_le.fit_transform(HF_balanced['activity'])
HF_balanced['activity'].value_counts()
```

```python
human_fall_X = HF_balanced.drop('activity',axis=1)
human_fall_y = HF_balanced['activity']

## normalizing the data
HF_scaler = StandardScaler()
X_scaled = HF_scaler.fit_transform(human_fall_X)
X_scaled = pd.DataFrame(X_scaled, columns=human_fall_X.columns)
X_scaled.head()
```

```python
# splitting the training data 80% and testing data 20%
HF_X_train, HF_X_test, HF_y_train, HF_y_test = train_test_split(X_scaled, human_fall_y, test_size=0.2, random_state=41)
print("Training HF shape 80%:", HF_X_train.shape, HF_y_train.shape)
print("Testing HF shape 20% :", HF_X_test.shape, HF_y_test.shape)
```

## Implementing Base Models





### ML Model SVM

```python
from sklearn.model_selection import GridSearchCV
from sklearn.svm import SVC
from sklearn.neighbors import KNeighborsClassifier
from sklearn.tree import DecisionTreeClassifier

from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import GradientBoostingClassifier
from sklearn.neural_network import MLPClassifier

from sklearn.metrics import classification_report, confusion_matrix, ConfusionMatrixDisplay
import time

params_svm = {
    'C': [0.1, 0.01, 2],
    'kernel': ['rbf', 'linear', 'lbfgs'],
    'gamma': ['scale', 'auto'],
    'degree': [1, 2, 3]
}

svm_model = SVC()
grid_svm = GridSearchCV(svm_model, params_svm, cv=2)
grid_svm.fit(HF_X_train.sample(1500), HF_y_train.sample(1500))
print("Best SVM parameters:", grid_svm.best_params_)

start_time = time.time()
svm_mod = grid_svm.best_estimator_
svm_mod.fit(HF_X_train, HF_y_train)
end_time = time.time()
print(f"\nSVM Training time: {end_time - start_time:.4f} seconds")


start_time = time.time()
HF_y_pred = svm_mod.predict(HF_X_test)
print(classification_report(HF_y_test, HF_y_pred))
end_time = time.time()
print(f"\nSVM Testing time: {end_time - start_time:.4f} seconds")

HF_confmat = confusion_matrix(HF_y_test, HF_y_pred)
ConfusionMatrixDisplay(confusion_matrix=HF_confmat, display_labels=[0, 1, 2, 3, 4, 5]).plot(cmap='crest')
plt.title('\nConfusion Matrix SVM')
plt.show()
```

### ML Model KNN

```python
params_knn = {
    'n_neighbors': [2, 3, 5, 7],
    'weights': ['uniform', 'distance'],
    'algorithm': ['auto', 'ball_tree', 'kd_tree', 'brute'],
    'p': [1, 2]
}

knn_model = KNeighborsClassifier()
grid_knn = GridSearchCV(knn_model, params_knn, cv=2)
grid_knn.fit(HF_X_train.sample(1500), HF_y_train.sample(1500))
print("Best KNN parameters:", grid_knn.best_params_)

start_time = time.time()
knn_mod = grid_knn.best_estimator_
knn_mod.fit(HF_X_train, HF_y_train)
end_time = time.time()
print(f"\nKNN Training time: {end_time - start_time:.4f} seconds")

start_time = time.time()
HF_y_pred = knn_mod.predict(HF_X_test)
print(classification_report(HF_y_test, HF_y_pred))
end_time = time.time()
print(f"\nKNN Testing time: {end_time - start_time:.4f} seconds")

HF_confmat = confusion_matrix(HF_y_test, HF_y_pred)
ConfusionMatrixDisplay(confusion_matrix=HF_confmat, display_labels=[0, 1, 2, 3, 4, 5]).plot(cmap='crest')
plt.title('\nConfusion Matrix KNN')
plt.show()
```

### ML Model DT


```python
params_dt = {
    'max_depth': [5, 10, None],
    'criterion': ['gini', 'entropy'],
    'min_samples_split': [2, 5, 10],
    'min_samples_leaf': [1, 2, 4]
}

dt_model = DecisionTreeClassifier()
grid_dt = GridSearchCV(dt_model, params_dt, cv=2)
grid_dt.fit(HF_X_train.sample(1500), HF_y_train.sample(1500))
print("Best DT parameters:", grid_dt.best_params_)

start_time = time.time()
dt_mod = grid_dt.best_estimator_
dt_mod.fit(HF_X_train, HF_y_train)
end_time = time.time()
print(f"\nDT Training time: {end_time - start_time:.4f} seconds")

start_time = time.time()
HF_y_pred = dt_mod.predict(HF_X_test)
print(classification_report(HF_y_test, HF_y_pred))
end_time = time.time()
print(f"\nDT Testing time: {end_time - start_time:.4f} seconds")

HF_confmat = confusion_matrix(HF_y_test, HF_y_pred)
ConfusionMatrixDisplay(confusion_matrix=HF_confmat, display_labels=[0, 1, 2, 3, 4, 5]).plot(cmap='crest')
plt.title('\nConfusion Matrix DT')
plt.show()

```

### ML Model LR

```python
params_lr = {
    'penalty': ['l1', 'l2', 'elasticnet'],
    'max_iter': [10, 50, 100],
    'solver': ['liblinear', 'lbfgs', 'newton-cg', 'sag'],
    'C': [0.1, 0.2, 0.4]
}

lr_model = LogisticRegression()
grid_lr = GridSearchCV(lr_model, params_lr, cv=2)
grid_lr.fit(HF_X_train.sample(1500), HF_y_train.sample(1500))
print("Best LR parameters:", grid_lr.best_params_)

start_time = time.time()
lr_mod = grid_lr.best_estimator_
lr_mod.fit(HF_X_train, HF_y_train)
end_time = time.time()
print(f"\nLR Training time: {end_time - start_time:.4f} seconds")

start_time = time.time()
HF_y_pred = lr_mod.predict(HF_X_test)
print(classification_report(HF_y_test, HF_y_pred))
end_time = time.time()
print(f"\nLR Testing time: {end_time - start_time:.4f} seconds")

HF_confmat = confusion_matrix(HF_y_test, HF_y_pred)
ConfusionMatrixDisplay(confusion_matrix=HF_confmat, display_labels=[0, 1, 2, 3, 4, 5]).plot(cmap='crest')
plt.title('\nConfusion Matrix LR')
plt.show()
```

### ML Model GB

```python
params_gb = {
    'loss': ['log_loss', 'exponential'],
    'criterion': ['friedman_mse', 'friedman_mse'],
    'learning_rate': [0.1, 0.2, 0.5, 1.0],
    'n_estimators': [10, 20, 40]
}

gb_model = GradientBoostingClassifier()
grid_gb = GridSearchCV(gb_model, params_gb, cv=2)
grid_gb.fit(HF_X_train.sample(1500), HF_y_train.sample(1500))
print("Best GB parameters:", grid_gb.best_params_)

start_time = time.time()
gb_mod = grid_gb.best_estimator_
gb_mod.fit(HF_X_train, HF_y_train)
end_time = time.time()
print(f"\nGB Training time: {end_time - start_time:.4f} seconds")

start_time = time.time()
HF_y_pred = gb_mod.predict(HF_X_test)
print(classification_report(HF_y_test, HF_y_pred))
end_time = time.time()
print(f"\nGB Testing time: {end_time - start_time:.4f} seconds")

HF_confmat = confusion_matrix(HF_y_test, HF_y_pred)
ConfusionMatrixDisplay(confusion_matrix=HF_confmat, display_labels=[0, 1, 2, 3, 4, 5]).plot(cmap='crest')
plt.title('\nConfusion Matrix GB')
plt.show()
```

### ML Model MLP

```python
params_mlp = {
    'hidden_layer_sizes': [30, 100, 20],
    'activation': ['relu', 'selu', 'tanh'],
    'solver': ['adam', 'lbfgs', 'rmsprop'],
    'batch_size': [10, 20, 40]
}

mlp_model = MLPClassifier()
grid_mlp = GridSearchCV(mlp_model, params_mlp, cv=2)
grid_mlp.fit(HF_X_train.sample(1500), HF_y_train.sample(1500))
print("Best MLP parameters:", grid_mlp.best_params_)

start_time = time.time()
mlp_mod = grid_mlp.best_estimator_
mlp_mod.fit(HF_X_train, HF_y_train)
end_time = time.time()
print(f"\nMLP Training time: {end_time - start_time:.4f} seconds")

start_time = time.time()
HF_y_pred = mlp_mod.predict(HF_X_test)
print(classification_report(HF_y_test, HF_y_pred))
end_time = time.time()
print(f"\nMLP Testing time: {end_time - start_time:.4f} seconds")

HF_confmat = confusion_matrix(HF_y_test, HF_y_pred)
ConfusionMatrixDisplay(confusion_matrix=HF_confmat, display_labels=[0, 1, 2, 3, 4, 5]).plot(cmap='crest')
plt.title('\nConfusion Matrix MLP')
plt.show()
```

### Combined using Stacking Classifier with Logistic Regression

```python
from sklearn.ensemble import StackingClassifier

#======================= Meta-Classifier (Stacking) =======================#

base_estimators = [
    ('svm', grid_svm.best_estimator_),
    ('knn', grid_knn.best_estimator_),
    ('dt', grid_dt.best_estimator_)
]

meta_classifier = LogisticRegression()
stack_model = StackingClassifier(estimators=base_estimators, final_estimator=meta_classifier, cv=2, n_jobs=-1)

start_time = time.time()
stack_model.fit(HF_X_train, HF_y_train)
end_time = time.time()
print(f"\nTraining Time: {end_time - start_time:.4f} sec")

start_time = time.time()
HF_y_pred = stack_model.predict(HF_X_test)
print(classification_report(HF_y_test, HF_y_pred))
end_time = time.time()
print(f"\nTesting Time: {end_time - start_time:.4f} sec")

# Confusion Matrix (for Stacking)
stack_confmat = confusion_matrix(HF_y_test, HF_y_pred)
ConfusionMatrixDisplay(confusion_matrix=stack_confmat, display_labels=stack_model.classes_).plot(cmap='Blues')
plt.title('\nCM of SVM, KNN, DT using Meta Classifier with LR')
plt.show()
```
